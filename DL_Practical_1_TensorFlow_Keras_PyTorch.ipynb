## 2. Build a simple Keras neural network

```python
# Import Keras from TensorFlow.
from tensorflow import keras

# Import Sequential to arrange layers one after another.
from tensorflow.keras.models import Sequential

# Import Dense for fully connected neural-network layers.
from tensorflow.keras.layers import Dense

# Create a neural network with 10 input features, 64 hidden neurons, and 1 output.
model = Sequential([
    Dense(64, activation='relu', input_shape=(10,)),  # Hidden layer: learns patterns from 10 input values.
    Dense(1, activation='sigmoid')                    # Output layer: returns a value between 0 and 1.
])
```

```python
# Configure how the model learns and how its performance is measured.
model.compile(
    optimizer='adam',             # Adam updates model weights during training.
    loss='binary_crossentropy',   # Loss function for a two-class prediction.
    metrics=['accuracy']          # Report the fraction of correct predictions.
)
```

```python
# Display the layers, output shapes, and parameter counts.
model.summary()
```

## 3. Install and use PyTorch

```python
# Import sys so pip runs through the Python interpreter used by this notebook.
import sys

# Install PyTorch and its vision/audio companion packages in this environment.
!{sys.executable} -m pip install torch torchvision torchaudio
```

```python
# Import PyTorch and its neural-network layer collection.
import torch
import torch.nn as nn

# Create the same simple network using PyTorch.
model_torch = nn.Sequential(
    nn.Linear(10, 64),  # Fully connected layer: 10 input features to 64 outputs.
    nn.ReLU(),           # Apply ReLU activation.
    nn.Linear(64, 1),   # Convert 64 hidden values into one output.
    nn.Sigmoid()         # Convert output into a value between 0 and 1.
)
```

```python
# Generate 5 sample rows, each containing 10 random input values.
x = torch.randn(5, 10)

# Pass the input through the network (forward propagation).
output = model_torch(x)

# Display the 5 predicted outputs.
print(output)
```

## 4. TensorFlow tensor addition and versions

```python
# Import TensorFlow with the common short name tf.
import tensorflow as tf

# Create two constant TensorFlow values.
a = tf.constant(10)
b = tf.constant(20)

# Add the two tensors and store the result.
c = tf.add(a, b)

# Convert the TensorFlow result to a regular NumPy/Python value and print it.
print(c.numpy())  # Expected output: 30
```
