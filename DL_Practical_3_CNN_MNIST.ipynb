# Practical 3 – Handwritten Digit Recognition using CNN

**Aim:** Build and train a Convolutional Neural Network to classify MNIST handwritten digit images.

Every important code line is explained using `# comments` directly inside the code. Run the cells from top to bottom.

## 1. Import libraries and load MNIST

```python
# TensorFlow provides the dataset and deep-learning operations.
import tensorflow as tf

# Matplotlib displays sample images and training graphs.
import matplotlib.pyplot as plt

# Select the built-in MNIST handwritten digit dataset.
mnist = tf.keras.datasets.mnist

# Load training images/labels and testing images/labels.
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Scale all pixel values from 0–255 into the 0–1 range.
x_train = x_train / 255.0
x_test = x_test / 255.0
```

## 2. Display sample images

```python
# Create a figure to hold the sample images.
plt.figure(figsize=(5, 3))

# Repeat the display steps for the first two training images.
for i in range(2):
    plt.subplot(1, 2, i + 1)             # Make a 1-row, 2-column layout.
    plt.imshow(x_train[i], cmap='gray')  # Display the image in grayscale.
    plt.title("Label: " + str(y_train[i]))  # Show the correct digit label.
    plt.axis('off')                      # Hide coordinate axes.

# Render the figure.
plt.show()
```

## 3. Import CNN layers and build model

```python
# Import the model container and layers needed for a CNN.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, Conv2D, Reshape, MaxPooling2D

# Create the CNN layer by layer.
model = Sequential([
    Input(shape=(28, 28)),                    # Input is a 28x28 grayscale image.
    Reshape((28, 28, 1)),                     # Add one channel dimension for grayscale.
    Conv2D(32, (3, 3), activation='relu'),    # Apply 32 filters of size 3x3 to learn image features.
    MaxPooling2D((2, 2)),                    # Reduce feature-map width and height by taking maxima.
    Flatten(),                               # Convert feature maps into one long vector.
    Dense(128, activation='relu'),            # Learn combinations of extracted features.
    Dense(10, activation='softmax')           # Return probabilities for the 10 digit classes.
])

# Display the CNN architecture.
model.summary()
```

## 4. Compile and train CNN

```python
# Configure the learning algorithm, classification loss, and accuracy metric.
model.compile(
    optimizer='adam',                         # Adam adjusts weights during training.
    loss='sparse_categorical_crossentropy',   # Loss for integer labels across 10 classes.
    metrics=['accuracy']                     # Track classification accuracy.
)
```

```python
# Train the CNN for 5 epochs and check it against the test set after each epoch.
history = model.fit(
    x_train,                    # Training images.
    y_train,                    # Correct digit labels.
    epochs=5,                    # Number of complete passes over training data.
    validation_data=(x_test, y_test)  # Validation images and labels.
)
```

## 5. Evaluate the CNN

```python
# Calculate loss and accuracy on the test dataset.
loss, accuracy = model.evaluate(x_test, y_test)

# Print the final evaluation values.
print("Loss:", loss)
print("Accuracy:", accuracy)
```

## 6. Plot training history

```python
# Draw training loss for each epoch.
plt.plot(history.history['loss'], label='loss')

# Draw validation loss for each epoch.
plt.plot(history.history['val_loss'], label='val_loss')

# Add labels and show the loss graph.
plt.title('Training Loss')
plt.xlabel('epochs')
plt.ylabel('loss')
plt.legend()
plt.show()
```

```python
# Draw training accuracy for each epoch.
plt.plot(history.history['accuracy'], label='accuracy')

# Draw validation accuracy for each epoch.
plt.plot(history.history['val_accuracy'], label='val_accuracy')

# Add labels and show the accuracy graph.
plt.title('Training Accuracy')
plt.xlabel('epochs')
plt.ylabel('accuracy')
plt.legend()
plt.show()
```
