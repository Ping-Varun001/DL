# Practical 2 – Handwritten Digit Recognition using ANN

**Aim:** Build and train an Artificial Neural Network on MNIST handwritten digit images and evaluate its accuracy.

Every important code line is explained using `# comments` directly inside the code. Run the cells from top to bottom.

```python
# Import TensorFlow for loading data and building the neural network.
import tensorflow as tf

# Import Sequential to stack neural-network layers in order.
from tensorflow.keras.models import Sequential

# Flatten converts images to vectors; Dense creates fully connected layers.
from tensorflow.keras.layers import Flatten, Dense

# Matplotlib is used to draw training and validation graphs.
import matplotlib.pyplot as plt
```

## 2. Load and prepare MNIST

```python
# Select the built-in MNIST handwritten digit dataset.
mnist = tf.keras.datasets.mnist

# Load images and labels, already split into training and testing sets.
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Scale pixel values from 0–255 to 0–1 for easier neural-network training.
x_train = x_train / 255.0
x_test = x_test / 255.0
```

## 3. Create the ANN

```python
# Create a feed-forward neural network with layers executed in sequence.
model = Sequential([
    Flatten(input_shape=(28, 28)),       # Convert each 28x28 image into 784 numbers.
    Dense(256, activation='relu'),       # First hidden layer with 256 neurons.
    Dense(128, activation='relu'),       # Second hidden layer with 128 neurons.
    Dense(10, activation='softmax')      # 10 output probabilities for digits 0 through 9.
])

# Print the network structure and parameter counts.
model.summary()
```

## 4. Compile and train

```python
# Set the optimizer, loss function, and metric before training.
model.compile(
    optimizer='sgd',                              # Stochastic Gradient Descent updates weights.
    loss='sparse_categorical_crossentropy',       # Multi-class loss when labels are integer values.
    metrics=['accuracy']                          # Track the percentage of correct predictions.
)
```

```python
# Train using training images and their correct digit labels.
history = model.fit(
    x_train,                         # Input training images.
    y_train,                         # Correct labels for those images.
    epochs=5,                         # Pass through the complete training set 5 times.
    validation_data=(x_test, y_test)  # Check performance on test images after each epoch.
)
```

## 5. Evaluate and print results

```python
# Evaluate the trained model on the test images it did not train on.
loss, accuracy = model.evaluate(x_test, y_test)

# Display the final test loss and accuracy.
print('loss:', loss)
print('accuracy:', accuracy)
```

## 6. Plot loss and accuracy

```python
# Plot loss recorded on the training data at each epoch.
plt.plot(history.history['loss'], label='loss')

# Plot loss recorded on validation/test data at each epoch.
plt.plot(history.history['val_loss'], label='val_loss')

# Add graph title, axis labels, legend, and display it.
plt.title("Training Loss")
plt.xlabel('epochs')
plt.ylabel('loss')
plt.legend()
plt.show()
```

```python
# Plot training accuracy at each epoch.
plt.plot(history.history['accuracy'], label='accuracy')

# Plot validation/test accuracy at each epoch.
plt.plot(history.history['val_accuracy'], label='val_accuracy')

# Add graph title, axis labels, legend, and display it.
plt.title("Training Accuracy")
plt.xlabel('epochs')
plt.ylabel('accuracy')
plt.legend()
plt.show()
```
